### Importação dos dados



```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_1.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_4.csv"

loja = pd.read_csv(url)
loja2 = pd.read_csv(url2)
loja3 = pd.read_csv(url3)
loja4 = pd.read_csv(url4)

loja.head()
```

#1. Análise do faturamento


Nesta primeira análise, você deve calcular o faturamento total de cada loja. Somando os valores da coluna Preço de cada loja para estimar o faturamento.

```python
# prompt: você deve calcular o faturamento total de cada loja. Somando os valores da coluna Preço de cada loja para estimar o faturamento.

faturamento_loja1 = loja['Preço'].sum()
faturamento_loja2 = loja2['Preço'].sum()
faturamento_loja3 = loja3['Preço'].sum()
faturamento_loja4 = loja4['Preço'].sum()


print(f" Faturamento Loja 1: R${faturamento_loja1:,.2f}, \n Faturamento Loja 2: R${faturamento_loja2:,.2f}, \n Faturamento Loja 3: R${faturamento_loja3:,.2f}, \n Faturamento Loja 4: R${faturamento_loja4:,.2f}")
```

```python
# prompt: calcular o ticket médio por loja

import pandas as pd

# Cálculo do ticket médio por loja
ticket_medio_loja1 = faturamento_loja1 / len(loja)
ticket_medio_loja2 = faturamento_loja2 / len(loja2)
ticket_medio_loja3 = faturamento_loja3 / len(loja3)
ticket_medio_loja4 = faturamento_loja4 / len(loja4)

print(f"\nTicket Médio Loja 1: R${ticket_medio_loja1:,.2f}")
print(f"Ticket Médio Loja 2: R${ticket_medio_loja2:,.2f}")
print(f"Ticket Médio Loja 3: R${ticket_medio_loja3:,.2f}")
print(f"Ticket Médio Loja 4: R${ticket_medio_loja4:,.2f}")
```

```python
# prompt: gráfico de barras com faturamento por loja e mostrar os valores

import matplotlib.pyplot as plt

# Dados de faturamento (substitua pelos seus valores reais)
lojas = ['Loja 1', 'Loja 2', 'Loja 3', 'Loja 4']
faturamentos = [faturamento_loja1, faturamento_loja2, faturamento_loja3, faturamento_loja4]

# Criar o gráfico de barras
plt.figure(figsize=(10, 6))  # Ajustar o tamanho da figura
plt.bar(lojas, faturamentos, color=['skyblue', 'salmon', 'lightgreen', 'lightcoral'])
plt.xlabel('Lojas')
plt.ylabel('Faturamento (R$)')
plt.title('Faturamento por Loja')

# Adicionar os valores de faturamento em cima de cada barra
for i, v in enumerate(faturamentos):
    plt.text(i, v, f'R${v:,.2f}', ha='center', va='bottom')

plt.show()
```

```python
# prompt: fazer um gráfico de linhas com os valores de faturamento por mês e ano

import matplotlib.pyplot as plt

loja['Data da Compra'] = pd.to_datetime(loja['Data da Compra'], format='%d/%m/%Y')
loja2['Data da Compra'] = pd.to_datetime(loja2['Data da Compra'], format='%d/%m/%Y')
loja3['Data da Compra'] = pd.to_datetime(loja3['Data da Compra'], format='%d/%m/%Y')
loja4['Data da Compra'] = pd.to_datetime(loja4['Data da Compra'], format='%d/%m/%Y')

# Criando colunas para mês e ano
loja['Mês'] = loja['Data da Compra'].dt.month
loja['Ano'] = loja['Data da Compra'].dt.year
loja2['Mês'] = loja2['Data da Compra'].dt.month
loja2['Ano'] = loja2['Data da Compra'].dt.year
loja3['Mês'] = loja3['Data da Compra'].dt.month
loja3['Ano'] = loja3['Data da Compra'].dt.year
loja4['Mês'] = loja4['Data da Compra'].dt.month
loja4['Ano'] = loja4['Data da Compra'].dt.year

# Agrupando por mês e ano e somando o faturamento
faturamento_mensal_loja1 = loja.groupby(['Ano', 'Mês'])['Preço'].sum()
faturamento_mensal_loja2 = loja2.groupby(['Ano', 'Mês'])['Preço'].sum()
faturamento_mensal_loja3 = loja3.groupby(['Ano', 'Mês'])['Preço'].sum()
faturamento_mensal_loja4 = loja4.groupby(['Ano', 'Mês'])['Preço'].sum()

# Convert the index to datetime objects for plotting
faturamento_mensal_loja1.index = pd.to_datetime(faturamento_mensal_loja1.index.map(lambda x: f"{x[0]}-{x[1]:02}"))
faturamento_mensal_loja2.index = pd.to_datetime(faturamento_mensal_loja2.index.map(lambda x: f"{x[0]}-{x[1]:02}"))
faturamento_mensal_loja3.index = pd.to_datetime(faturamento_mensal_loja3.index.map(lambda x: f"{x[0]}-{x[1]:02}"))
faturamento_mensal_loja4.index = pd.to_datetime(faturamento_mensal_loja4.index.map(lambda x: f"{x[0]}-{x[1]:02}"))


# Plotando o gráfico de linhas para a Loja 1
plt.figure(figsize=(12, 6))
plt.plot(faturamento_mensal_loja1.index, faturamento_mensal_loja1.values, label='Loja 1', marker='o')
plt.plot(faturamento_mensal_loja2.index, faturamento_mensal_loja2.values, label='Loja 2', marker='o')
plt.plot(faturamento_mensal_loja3.index, faturamento_mensal_loja3.values, label='Loja 3', marker='o')
plt.plot(faturamento_mensal_loja4.index, faturamento_mensal_loja4.values, label='Loja 4', marker='o')

plt.xlabel('Ano e Mês')
plt.ylabel('Faturamento (R$)')
plt.title('Faturamento Mensal por Loja')
plt.xticks(rotation=45)
plt.legend()
plt.tight_layout()
plt.show()
```

# 2. Vendas por Categoria



Neste passo, deve calcular a quantidade de produtos vendidos por categoria em cada loja. A ideia é agrupar os dados por categoria e contar o número de vendas de cada tipo, mostrando as categoria

```python
# prompt: faturamento por Categoria de produto

import matplotlib.pyplot as plt

# Calcula a quantidade de produtos vendidos por categoria em cada loja
vendas_por_categoria_loja1 = loja['Categoria do Produto'].value_counts()
vendas_por_categoria_loja2 = loja2['Categoria do Produto'].str.strip().value_counts()
vendas_por_categoria_loja3 = loja3['Categoria do Produto'].str.strip().value_counts()
vendas_por_categoria_loja4 = loja4['Categoria do Produto'].str.strip().value_counts()

# Criando um DataFrame com os dados de vendas por categoria para todas as lojas
vendas_categoria = pd.concat([vendas_por_categoria_loja1,
                              vendas_por_categoria_loja2,
                              vendas_por_categoria_loja3,
                              vendas_por_categoria_loja4], axis=1)

# Renomeando as colunas para melhor visualização
vendas_categoria.columns = ['Loja 1', 'Loja 2', 'Loja 3', 'Loja 4']

# Exibindo o DataFrame com as vendas por categoria
print(vendas_categoria)

# Plotando o gráfico de barras para vendas por categoria em cada loja
vendas_categoria.plot(kind='bar', figsize=(10, 6))
plt.xlabel('Categoria do Produto')
plt.ylabel('Quantidade Vendida')
plt.title('Vendas por Categoria de Produto em Cada Loja')
plt.xticks(rotation=45, ha='right')  # Rotaciona os rótulos do eixo x para melhor legibilidade
plt.legend(title='Lojas')
plt.tight_layout()  # Ajusta o layout para evitar sobreposição de elementos
plt.show()
```

# 3. Média de Avaliação das Lojas

Neste passo, vamos calcular a média das avaliações dos clientes para cada loja. O objetivo é entender a satisfação dos clientes com os produtos vendidos.

```python
# Calcula a média das avaliações dos clientes para cada loja
avaliacao_media_loja1 = loja['Avaliação da compra'].mean()
avaliacao_media_loja2 = loja2['Avaliação da compra'].mean()
avaliacao_media_loja3 = loja3['Avaliação da compra'].mean()
avaliacao_media_loja4 = loja4['Avaliação da compra'].mean()

print(f"\nAvaliação média da Loja 1: {avaliacao_media_loja1:.2f}")
print(f"Avaliação média da Loja 2: {avaliacao_media_loja2:.2f}")
print(f"Avaliação média da Loja 3: {avaliacao_media_loja3:.2f}")
print(f"Avaliação média da Loja 4: {avaliacao_media_loja4:.2f}")
```

# 4. Produtos Mais e Menos Vendidos

Neste passo, deve identificar quais produtos foram os mais vendidos e os menos vendidos em cada loja.  Ao final, visualize os resultados de forma que fique claro quais produtos se destacaram nas vendas de cada loja.

```python
# 4. Produtos Mais e Menos Vendidos

def analyze_sales(df, loja_num):
    vendas_produto = df['Produto'].value_counts()
    mais_vendidos = vendas_produto.head(5)
    menos_vendidos = vendas_produto.tail(5)
    print(f"\nLoja {loja_num}:")
    print("Produtos Mais Vendidos:")
    print(mais_vendidos)
    print("\nProdutos Menos Vendidos:")
    print(menos_vendidos)

analyze_sales(loja, 1)
analyze_sales(loja2, 2)
analyze_sales(loja3, 3)
analyze_sales(loja4, 4)
```

# 5. Frete Médio por Loja


Neste passo, vamos calcular o custo médio de frete para cada loja. O objetivo é entender quanto, em média, está sendo gasto com frete para cada uma das lojas.

```python
# Calcula o custo médio de frete para cada loja
frete_medio_loja1 = loja['Frete'].mean()
frete_medio_loja2 = loja2['Frete'].mean()
frete_medio_loja3 = loja3['Frete'].mean()
frete_medio_loja4 = loja4['Frete'].mean()

print(f"\nFrete médio da Loja 1: R${frete_medio_loja1:,.2f}")
print(f"Frete médio da Loja 2: R${frete_medio_loja2:,.2f}")
print(f"Frete médio da Loja 3: R${frete_medio_loja3:,.2f}")
print(f"Frete médio da Loja 4: R${frete_medio_loja4:,.2f}")
```
